# MATPLOTLIB

#### Matplotlib is Python's most widely used data visualisation library. It produces publication-quality plots in a variety of formats and can be embedded in applications.

#### Why Matplotlib?
#### It is the foundation of other plotting libraries like Seaborn and Pandas .plot(). Full control over every element of a figure — fonts, colours, axes, ticks, gridlines. Works in scripts, Jupyter notebooks, and GUI applications.

#### Matplotlib Architecture

Layer: Role

* Figure: The top-level container — the entire window or canvas
* Axes: An individual plot area inside the Figure (has x-axis, y-axis, title)
* Artist: Everything drawn on the canvas: lines, text, patches, ticks, labels

#### Figure vs Axes
* One Figure can contain multiple Axes (subplots). 
* plt.figure() creates the Figure. 
* plt.axes() or plt.subplots() creates the Axes.
* Most plt.xxx() calls act on the CURRENT active Axes automatically.

#### Two Ways to Plot
* pyplot (plt.plot): Quick, interactive — great for simple single plots
* OO (fig, ax = plt.subplots()): Recommended for multi-plot figures and customisation

#### Important Methods
* plt.figure(figsize): Create figure with given size (inches)
* plt.subplots(r, c): Create figure + grid of r×c axes
* Syntax: fig, ax = plt.subplots(nrows=1, ncols=1, sharex=False, sharey=False, figsize=(width, height), **kwargs)

* plt.tight_layout(): Auto-adjust spacing between subplots
* plt.savefig(path): Save figure to file
* plt.clf(): Clear current figure
* plt.close(): Close current figure window

```python
# Subplots — Multiple Plots in One Figure
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8))
axes[0,0].plot([1,2,3], [4,5,6])
axes[0,0].set_title('Top Left')
axes[0,1].scatter([1,2,3], [6,2,8])
axes[0,1].set_title('Top Right')
axes[1,0].bar(['A','B','C'], [3,7,5])
axes[1,0].set_title('Bottom Left')
axes[1,1].hist([1,2,2,3,3,3,4,4,4,4])
axes[1,1].set_title('Bottom Right')
plt.tight_layout() # prevent overlap
plt.show()
```

#### Line Style & Marker Reference
* color: 'blue' | 'red' | '#FF5733' | (0.1, 0.2, 0.5) — RGB tuple
* linestyle: '-' solid | '--' dashed | ':' dotted | '-.' dash-dot
* inewidth: Float — e.g. 1.5, 2.0, 3.0

* markersize: Float — size of the marker
* alpha: 0.0 (transparent) to 1.0 (opaque) — controls transparency
* label: String shown in legend

```python
# Line Styling
x = np.linspace(0, 2 * np.pi, 100)
plt.plot(x, np.sin(x), color='blue', linestyle='-', linewidth=2, label='sin(x)')
plt.plot(x, np.cos(x), color='red', linestyle='--', linewidth=2, label='cos(x)')
plt.plot(x, np.tan(x), color='green', linestyle=':', linewidth=1, label='tan(x)')
plt.title('Trigonometric Functions')
plt.xlabel('Angle (radians)')
plt.ylabel('Value')
plt.ylim(-2, 2) # limit y-axis range
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Axis Customisation
x = [1, 2, 3, 8, 5]
y = [2, 4, 1, 4, 0]
plt.plot(x, y)
plt.xlim(0, 10) # set x-axis range
plt.ylim(-5, 5) # set y-axis range
plt.xticks([0, 2, 4, 6, 8, 10]) # set x tick positions
plt.yticks([-4, -2, 0, 2, 4]) # set y tick positions
plt.xticks(rotation=45) # rotate x labels
# Add annotations
plt.annotate('Peak', xy=(1.57, 1),
xytext=(2.5, 0.8),
arrowprops=dict(arrowstyle='->'))
# Add horizontal / vertical reference lines
plt.axhline(y=0, color='black', linewidth=0.8) # horizontal
plt.axvline(x=np.pi, color='grey', linestyle='--') # vertical
```

#### plt.figure() vs plt.subplots()
• plt.figure(): Creates only the blank window/canvas. You must add axes or plots to it manually.

• plt.subplots(): A modern convenience function that creates both the figure canvas and one or more grids of axes at the exact same time. (fig, ax = plt.subplots())

```python
fig = plt.figure(figsize=(6, 4), facecolor='lightgray')

# Add an axes mapping: (left, bottom, width, height) as a fraction of the figure
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])
ax.plot([0, 1, 2], [0, 1, 4])

plt.show()
```

#### What is fig.add_axes()?
#### fig.add_axes() manually creates a new plotting area (an Axes object) inside your blank canvas and gives you total control over its exact size and position.
 
#### How the Coordinates Work:
The method takes a list of four numbers: [left, bottom, width, height].

These numbers are fractions of the total figure size (ranging from 0.0 to 1.0):
• left (0.1): The left side of the plot starts at 10% of the way from the canvas's left edge.

• bottom (0.1): The bottom of the plot starts at 10% of the way from the canvas's bottom edge.

• width (0.8): The plot stretches horizontally to cover 80% of the canvas width.

• height (0.8): The plot stretches vertically to cover 80% of the canvas height.

This leaves a uniform 10% margin around all four sides of the plot, which provides just enough room for your axis ticks and labels.



```python
# 1. Generate some dummy data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# 2. Create the main canvas
fig = plt.figure(figsize=(8, 5))

# 3. Create the main full-sized plot
# [left, bottom, width, height] as fractions of the canvas
ax_main = fig.add_axes([0.1, 0.1, 0.8, 0.8])
ax_main.plot(x, y, color='blue', label='Main Sine Wave')
ax_main.set_title('Main Plot')
ax_main.set_xlabel('X Axis')
ax_main.set_ylabel('Y Axis')

# 4. Create an inset plot in the upper-right corner
# It starts at 60% left, 60% bottom, and takes up 25% width/height
ax_inset = fig.add_axes([0.6, 0.6, 0.25, 0.25])
ax_inset.plot(x[:30], y[:30], color='red')  # Zoomed into just the first 30 points
ax_inset.set_title('Zoomed Inset', fontsize=10)

plt.show()
```

#### Scatter Plot:
#### Scatter plots show the relationship between two continuous variables. Each point represents one observation. Ideal for spotting correlations, clusters, and outliers.

```python
np.random.seed(42)
x = np.random.randn(100)
y = 2 * x + np.random.randn(100) * 0.5 # positive correlation
plt.figure(figsize=(7, 5))
plt.scatter(x, y, color='steelblue', alpha=0.7, edgecolors='white', s=60)
plt.title('Scatter Plot — Positive Correlation')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```

#### Scatter Plot Parameters
* s: Marker size (scalar or array — vary by data value)
* c: Colour (string, array, or colormap-mapped values)
* cmap: Colormap when c is numeric: 'viridis' | 'plasma' | 'coolwarm'
* alpha: Transparency 0–1 (helps with overlapping points)
* edgecolors: Border colour of each marker
* marker: Shape: 'o' | '^' | 's' | 'D' | '*'

```python
# Colour-Mapped & Size-Mapped Scatter
# Bubble chart — size and colour encode extra dimensions
np.random.seed(10)
n = 50
x = np.random.rand(n)
y = np.random.rand(n)
sizes = np.random.rand(n) * 500 # bubble size
colors = np.random.rand(n) # colour value
plt.figure(figsize=(8, 6))
sc = plt.scatter(x, y, s=sizes, c=colors, cmap='viridis', alpha=0.7)
plt.colorbar(sc, label='Colour Scale') # add colour bar
plt.title('Bubble Chart')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

#### numpy.polyfit() or Polynomial:
#### It finds the best-fit polynomial equation for your data points—meaning it helps you approximate patterns and trends mathematically.
#### How Does numpy.polyfit() Work?

It’s super simple:

You give it x-values (independent variable).

You give it y-values (dependent variable).

You specify the degree of the polynomial you want.

And just like that, NumPy does the math and gives you the polynomial equation.

```python
## Adding a Trend Line
# Fit and plot a linear trend line
from numpy.polynomial import Polynomial

# Sample data
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([2, 3, 5, 7, 11, 13])

m, b = Polynomial.fit(x, y, deg=1) # degree-1 polynomial (line)
trend = m * x + b
plt.scatter(x, y, alpha=0.6, label='Data')
plt.plot(np.sort(x), m * np.sort(x) + b,
color='red', linewidth=2, label=f'y = {m:.2f}x + {b:.2f}')
plt.legend()
plt.title('Scatter with Trend Line')
plt.show()
```

```python
# Sample data points
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([2, 3, 5, 7, 11, 13])

# Fit a quadratic polynomial (degree = 2)
coefficients = np.polyfit(x, y, 2)

# Generate fitted y-values
y_fit = np.polyval(coefficients, x)

# Plot original data and the quadratic fit
plt.scatter(x, y, label='Original Data', color='red')
plt.plot(x, y_fit, label='Quadratic Fit', color='green')
plt.legend()
plt.show()
```

#### Bar Plot:
#### Bar plots compare discrete categories. Each bar's height (or length) represents a value. Best used when the x-axis is categorical, not continuous.

```python
categories = ['Python', 'JavaScript', 'Java', 'C++', 'Go']
popularity = [30.3, 13.5, 17.8, 10.2, 5.1]
plt.figure(figsize=(8, 5))
bars = plt.bar(categories, popularity, color='steelblue', edgecolor='navy',
width=0.6, label='%')
# Add value labels on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.3,
             f'{height}%', ha='center', va='bottom', fontsize=10)
plt.title('Programming Language Popularity')
plt.xlabel('Language')
plt.ylabel('Popularity (%)')
plt.legend()
plt.ylim(0, 35)
plt.tight_layout()
plt.show()
```

#### Bar Plot Parameters
* width: Width of bars (0–1, default 0.8)
* color: Fill colour — string, list, or hex
* edgecolor: Border colour of bars
* align: 'center' (default) | 'edge'
* bottom: Y value where bar starts — used for stacked bars
* label: Label for legend

#### When to use Bar vs Line:
* Bar plot: comparing discrete, unordered categories (languages, countries, departments).
* Line plot: showing a trend over ordered/continuous data (months, years, temperature over time).

#### Horizontal, Multiple & Stacked Bars:

#### Horizontal Bar Plot:

Use horizontal bars when category names are long or when comparing rankings feels more natural left-to-right.

#### Multiple (Grouped) Bar Plot:

Group multiple bars side-by-side to compare sub-categories across the same groups.

```python
subjects = ['Maths', 'Science', 'English', 'History']
student_A = [85, 92, 78, 88]
student_B = [76, 84, 90, 72]
student_C = [91, 88, 82, 95]

x = np.arange(len(subjects))
width = 0.25
plt.figure(figsize=(9, 5))
plt.bar(x - width, student_A, width, label='Alice', color='steelblue')
plt.bar(x, student_B, width, label='Bob', color='tomato')
plt.bar(x + width, student_C, width, label='Charlie', color='seagreen')
plt.xticks(x, subjects)
plt.xlabel('Subject')
plt.ylabel('Score')
plt.title('Student Score Comparison')
plt.legend()
plt.tight_layout()
plt.show()
```

#### Stacked Bar Plot:

Stack bars to show both the total and the composition of each category.

```python
quarters = ['Q1', 'Q2', 'Q3', 'Q4']
product_A = [20, 25, 30, 28]
product_B = [15, 18, s22, 20]
product_C = [10, 12, 14, 18]

plt.figure(figsize=(8, 5))
plt.bar(quarters, product_A, label='Product A', color='steelblue')
plt.bar(quarters, product_B, label='Product B', color='tomato',
bottom=product_A)
# For third layer, bottom = A + B
bottom_AB = [a+b for a,b in zip(product_A, product_B)]
plt.bar(quarters, product_C, label='Product C', color='seagreen',
bottom=bottom_AB)
plt.title('Quarterly Sales by Product (Stacked)')
plt.xlabel('Quarter')
plt.ylabel('Sales')
plt.legend()
plt.tight_layout()
plt.show()
```

#### Histogram:

A histogram shows the distribution of a continuous variable by grouping values into bins and counting how many fall in each bin. It answers: 'what does the data look like?'

#### Bar chart vs Histogram:
* Bar chart: categories on x-axis (discrete, unordered).
* Histogram: continuous numeric data on x-axis, grouped into bins.
* No gaps between bars in a histogram (continuous data).

```python
np.random.seed(0)
data = np.random.randn(1000) # 1000 values from normal distribution
plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, color='steelblue', edgecolor='white', alpha=0.8)
plt.title('Normal Distribution — 1000 Samples')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```

#### Histogram Parameters:
* bins: Number of bins (int) or bin edges (list). Default: 10
* range (min, max): — limit data range included
* density: True → y-axis shows probability density instead of count
* cumulative: True → each bar shows running total
* color: Fill colour of bars
* edgecolor: Border colour (use 'white' to visually separate bins)
* alpha: Transparency — useful when overlapping histograms
* label: Legend label

```python
# Overlapping Histograms & KDE
group_A = np.random.normal(60, 10, 500) # mean=60, std=10
group_B = np.random.normal(75, 8, 500) # mean=75, std=8
plt.figure(figsize=(9, 5))
plt.hist(group_A, bins=30, alpha=0.6, color='steelblue', label='Group A')
plt.hist(group_B, bins=30, alpha=0.6, color='tomato', label='Group B')
plt.title('Score Distribution: Group A vs Group B')
plt.xlabel('Score')
plt.ylabel('Count')
plt.legend()
plt.show()
# Add KDE curve (requires scipy)
from scipy.stats import gaussian_kde
kde = gaussian_kde(group_A)
x_range = np.linspace(group_A.min(), group_A.max(), 200)
plt.plot(x_range, kde(x_range) * len(group_A) * (group_A.max()-group_A.min())/30,
'navy', linewidth=2, label='KDE')
```

#### Choosing the Right Number of Bins:
* Sturges: bins = 1 + log2 (n) — good for symmetric data
* Square root: bins = √n — simple, general purpose
* Auto (NumPy): np.histogram_bin_edges(data, bins='auto') — picks best
* Too few bins: Hides the shape — distribution looks flat
* Too many bins: Noisy — every bar is 1–2 observations
