# NUMPY

#### Numpy is a fundamental package for scientific computing in python. It is a python library that provides a Multi Dimensional Array Object(ndarray) and various derived objects such as Masked Arrays or Matrices and an assortment of routines for fast operations on arrays including Mathematical, Logical, shape manipulation, Sorting, Selecting, I/O, Discrete Fourier Transform, Basic Linear Algebra, Statistical Operations, Radom Simulation etc.

#### At the core of the Numpy package is the ndimensional Array object. 

### Why NumPy over Python lists?
##### Python list: [1, 2, 3] stores pointers to Python objects — slow and memory-heavy. NumPy array: stores raw numbers in contiguous memory — up to 50× faster for large datasets.This speed is critical for ML/AI — every tensor in TensorFlow & PyTorch is built on NumPy concepts.

## Numpy Arrays Vs Python Lists

#### Numpy Arrays have a fixed size at creation, unlike Python Lists which grow dynamically. Changing the size of a ndim array will create a new array and delete the original.

#### The elements in a Numpy array are required to be of same data type and thus of the same size in memory.
#### Numpy Arrays facilitate advanced Mathematical and other types of operations on huge amount of data. Such operations are executed efficiently with less code.

### ravel() --> a function used to flatten a multi-dimensional array into a one-dimensional array. 
#### np.ravel() is that it returns a view of the original array whenever possible, rather than a copy. This makes it highly memory-efficient. However, because it returns a view, modifying elements in the raveled array will directly modify the original array.
### How it is different from flatten():
* ravel() returns a view of the original array, while flatten() returns a completely new copy of the data in memory.
*    ravel(): If you change a value in the flattened array, the original matrix changes too because they share memory.
*    flatten(): If you change a value in the flattened array, the original matrix stays completely safe and unchanged.
* ravel(): Faster and uses less computer memory because it does not duplicate data.
* flatten(): Slower and uses more memory because it has to copy every single item over to a new location.

```python
print(m.ravel())
```

```python
# Create a proper 2x2 matrix
m = np.array([[1, 2], [3, 4]])

# Using ravel (shares memory)
r = m.ravel()
r[0] = 99  # Change first element

print("Original m after ravel change:\n", m) 
# Output shows original changed: [[99  2] [ 3  4]]

# Reset m
m = np.array([[1, 2], [3, 4]])

# Using flatten (makes a copy)
f = m.flatten()
f[0] = 99  # Change first element

print("\nOriginal m after flatten change:\n", m) 
# Output shows original is safe: [[1 2] [3 4]]
```

### Stacking & Splitting

#### stacking combines multiple arrays into a single new array, while splitting breaks one single array into multiple smaller sub-arrays.

### Stacking(Joining Data):
* np.vstack(): Vertical Stack: Joins arrays row-wise (on top of each other). [Must have the same number of columns.]
* np.hstack(): Horizontal Stack: Joins arrays column-wise (side-by-side). [Must have the same number of rows.]
* np.stack(): New Axis Stack: Joins arrays along a completely new dimension. [Must have the exact same shape.]

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

## Stacking
print(np.vstack([a, b])) # vertical: [[1,2,3],[4,5,6]] # Creates a 2D array by stacking rows
print(np.hstack([a, b])) # horizontal: [1,2,3,4,5,6] # Appends elements side-by-side
print(np.concatenate([a, b])) # same as hstack for 1D

# New Axis Stacking (axis=0 is default) -> Adds a new dimension
stacked = np.stack((a, b), axis=1)
print(stacked)
print(stacked.shape)  # Output: (2, 3)
```

### Splitting Arrays (Dividing Data):
#### Splitting is the exact opposite of joining. It parses one array into a Python list of smaller sub-arrays.
* np.split(): Splits an array into equal-sized sub-arrays. If the elements cannot be divided equally by the number of requested splits, it raises a ValueError
* np.array_split(): Splits an array into unequal or equal parts. If the array doesn't divide evenly, it automatically adjusts and packs the extra elements into the first few sub-arrays without throwing an error.
* np.vsplit() and np.hsplit(): Specialized functions to split multi-dimensional arrays vertically (row-wise) or horizontally (column-wise).

```python
## Splitting
arr = np.array([1, 2, 3, 4, 5, 6, 7])

# Using array_split because 7 items cannot be divided evenly into 3 parts
sub_arrays = np.array_split(arr, 3)
print('array_split():', sub_arrays) # Output: [array([1, 2, 3]), array([4, 5]), array([6, 7])]

# Accessing individual parts using index notation
print(sub_arrays[0])  # Output: [1, 2, 3]

m = np.arange(16).reshape(4, 4)
print(m)
print(np.vsplit(m, 2)) # split into 2 equal row-halves
print(np.hsplit(m, 2)) # split into 2 equal col-halves
```

### Boolean & Fancy Indexing

#### Fancy Indexing allows us to use an array of indices to access multiple array elements at once.

```python
a = np.array([10, 25, 3, 47, 8, 33])
# Boolean mask
mask = a > 20
print(mask) # [False True False True False True]
print(a[mask]) # [25, 47, 33] — filter in one line
print(a[a % 2 == 0]) # [10, 8] — even numbers only

# Fancy indexing — pick by index list
print(a[[0, 2, 5]]) # [10, 3, 33]
# Fancy indexing 2-D Array
matrix = np.array([
    [ 0,  1,  2],
    [10, 11, 12],
    [20, 21, 22]
])

# We want elements at (0, 0), (1, 2), and (2, 1)
row_indices = [0, 1, 2]
col_indices = [0, 2, 1]

print(matrix[row_indices, col_indices]) # Output: [ 0, 12, 21]
print(matrix[[0, 2], :]) # 1st and 3rd row

# Modifying Values Using Fancy Indexing
arr = np.array([10, 20, 30, 40, 50])
arr[[1, 3]] = 99

print(arr) # Output: [10, 99, 30, 99, 50]
```

### Broadcasting Rules:

#### Broadcasting in NumPy allows us to perform arithmetic operations on arrays of different shapes without reshaping them. It automatically adjusts the smaller array to match the larger array's shape by replicating its values along the necessary dimensions.
#### Rules:
* Rule 1: If arrays differ in ndim, prepend 1s to the smaller shape
* Rule 2: Dimensions of size 1 are stretched to match the other
* Rule 3: If sizes don't match and neither is 1 → ValueError

```python
# Broadcasting a Scalar to a 1D Array
arr = np.array([1, 2, 3])
res = arr + 1  
print(res)

# Broadcasting a 1D Array to a 2D Array
a = np.array([2, 4, 6])
b = np.array([[1, 3, 5], [7, 9, 11]])
# NumPy automatically expands the 1D array along the rows of the 2D array 
# to perform element-wise addition.
res = a + b
print(res)

# Broadcasting in Conditional Operations
a = np.array([12, 24, 35, 45, 60, 72])
b = np.array(["Adult", "Minor"])
res = np.where(a > 18, b[0], b[1])
print(res)

# Broadcasting for Matrix Multiplication
m = np.array([[1, 2], [3, 4]]) # (2,1)
v = np.array([10, 20]) # (2,) -> (1,2)
res = m * v
print(res)

# Rule: 1
a = np.array([[1],[2],[3]]) # shape (3,1)
b = np.array([10, 20, 30]) # shape (3,) → treated as (1,3)
print(a + b)

# Rule 2: Dimensions of size 1 are stretched to match the other
# A matrix with 3 rows and 2 columns
A = np.array([[1, 2], [3, 4], [5, 6]])  # Shape: (3, 2)

# A matrix with 1 row and 2 columns
B = np.array([[10, 20]])                # Shape: (1, 2) -> (3,2)
print(A + B)

# Rule 3: If sizes don't match and neither is 1 → ValueError
A = np.ones((3, 2))  # Shape: (3, 2)
B = np.ones((3, 3))  # Shape: (3, 3)

## Comparing shapes from right to left:
## Axis 1: 2 vs 3 -> Neither is 1! NumPy doesn't know how to stretch a 2 into a 3.
# print(A + B) # ValueError
```
